<h1>Other Classifiers</h1>

Decision Trees and Random Forest are not the only classifier algorithms, although arguably the easiest to understand. This document describes a few more classifier models. <br>
For better comparability between the different models, the <i>iris</i> dataset will be used with all of them.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

<br><h2>1. Naive Bayes</h2>

<i>Naive Bayes</i> classifiers utilise Bayes' theorem as a formula to calculate the probability of a class based on a set of features. It is "naive" in the sense that the features are assumed to be independent of each other. There are multiple types of Naive Bayes classifiers with slightly different use-cases. The model presented here is a Gaussian Naive Bayes that assumes a normal distribution for all values of the different features.<br>
<h3>Pros:</h3>
<ol>
    <li>Efficient/fast to run</li>
    <li>Can handle larger amounts of features</li>
    <li>Good performance even with smaller training sets</li>
    <li>Works with categorical data</li>
</ol>
<h3>Cons:</h3>
<ol>
    <li>Assumes feature-independence, which is rarely the case with real world data</li>
    <li>Outcomes not present in the training data might not be recognised by the model</li>
    <li>Data should generally not contain any missing values</li>
</ol>

<h3>Implementation:</h3>

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model_GNB = GaussianNB() #Calling the model
model_GNB.fit(X_train, y_train) #Training the model
pred_GNB = model_GNB.predict(X_test) #Letting the model predict on the test set
```

```python
print('Testing accuracy: ', accuracy_score(y_test, pred_GNB))
print('Training accuracy: ', accuracy_score(y_train, model_GNB.predict(X_train)))
```

<br><h2>2. Support Vector Machine</h2>

<i>Support Vector Machines</i> (SVC in the case of classification) try to fit data into different categories by finding the best separating boundary between the samples. The data points closest to the boundary are the so-called <i>support vectors</i>, the boundary itself is called <i>hyperplane</i>. The dimensionality of this boundary can be defined when initialising by specifying the model's kernel. A linear kernel will separate classes using a straight line. Polynomial kernels increase the dimensionality and allow the classifier a more complex distinction between classes. Kernels can also be customised, allowing for a lot of flexibility depending on the needs for the present data. A visualisation of different kernels can be seen below.

<img src = 'https://scikit-learn.org/stable/_images/sphx_glr_plot_iris_svc_001.png' alt = 'Visualisation of a Support Vector classifier'>

<h3>Pros:</h3>
<ol>
    <li>Can handle high dimensional data, even when the amount of dimensions is greater than the number of samples</li>
    <li>Classification is less affected by outliers in the dataset</li>
    <li>High flexibility with custom kernels</li>
</ol>
<h3>Cons:</h3>
<ol>
    <li>Gives no probability estimate for specific samples, these have to be calculated separately</li>
    <li>Steep increase in computing power and memory with a higher number of support vectors</li>
</ol>

<h3>Implementation:</h3>

```python
from sklearn import svm
```

```python
linear_SVC = svm.LinearSVC() #LinearSVC is implemented differently than a SVC with (kernel = linear) and scales better with larger datasets
poly_SVC = svm.SVC(kernel = 'poly', degree = 5) #Using the degree parameter, you can specify the dimension of the hyperplane; Default is 3

linear_SVC.fit(X_train, y_train)
poly_SVC.fit(X_train, y_train) #fitting both models

pred_linSVC = linear_SVC.predict(X_test)
pred_polySVC = poly_SVC.predict(X_test)
```

```python
print('Linear Testing Accuracy: ', accuracy_score(y_test, pred_linSVC)) #calculating testing accuracy of the linear SVC
print('Linear Training Accuracy: ', accuracy_score(y_train, linear_SVC.predict(X_train))) #calculating training accuracy of the linear SVC

print('Polynomial Testing Accuracy: ', accuracy_score(y_test, pred_polySVC)) #testing accuracy of the polynomial SVC
print('Polynomial Training Accuracy: ', accuracy_score(y_train, poly_SVC.predict(X_train))) #training accuracy of the polynomial SVC
```

<br><h2>3. Multi-Layer Perceptron</h2>

<i>Multi-Layer Perceptrons</i> are a type of neural network model. As the name suggests, it consistsof multiple layers, each of which contain a number of nodes called neurons. The first layer is the <b>input layer</b>, its neurons each corresponding to a specific feature in the data. The last layer is the <b>output layer</b> that generates the final output based on the data it receives from the layers inbetween, called <b>hidden layers</b>. The output layer contains as many outputs as there are classes in the data. The amount of neurons in the hidden layers can be controlled via parameters.
<h3>Pros:</h3>
<ol>
    <li>Able to learn on highly non-linear data</li>
    <li>Works well even on large datasets</li>
</ol>
<h3>Cons:</h3>
<ol>
    <li>Sensitive to small changes in parameters, thus fine-tuning the model is important</li>
</ol>

<h3>Implementation:</h3>

```python
from sklearn.neural_network import MLPClassifier
```

```python
model_MLP = MLPClassifier(solver = 'lbfgs', hidden_layer_sizes = 10, max_iter = 500) #Initiating the model with the lbfgs solver
model_MLP.fit(X_train, y_train) #Training the model
pred_MLP = model_MLP.predict(X_test) #Testing the model
```

<ol>
    <li>By default, the classifier uses the <i>adam</i> solver that works well on large datasets. The <i>iris</i> dataset is rather small, so the <i>lbfgs</i> solver is the better choice in terms of performance and efficiency.</li>
    <li><i>max_iter</i>: During training, the model iterates a certain number of times, adjusting weight and bias of features to minimise error until it reaches either convergence (having learned the data) or the amount of iterations given by this parameter.<br>Default is <i>max_iter = 200</i></li>
    <li><i>hidden_layer_sizes</i>: Parameter controlling the amount of neurons in the hidden layers. Note that if the layers are too small, the model may reach the maximum amount of iterations and fail to converge.<br>Default is <i>hidden_layer_sizes = 100</i></li>
</ol>

```python
print('Testing Accuracy: ', accuracy_score(y_test, pred_MLP))
print('Training Accuracy: ', accuracy_score(y_train, model_MLP.predict(X_train)))
```
