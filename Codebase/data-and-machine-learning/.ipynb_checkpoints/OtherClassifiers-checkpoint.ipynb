<h1>Other Classifiers</h1>

Decision Trees and Random Forest are not the only classifier algorithms, although arguably the easiest to understand. This document describes a few more classifier models. <br>
For better comparability between the different models, the <i>iris</i> dataset will be used with all of them.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

<br><h2>1. Naive Bayes</h2>

<i>Naive Bayes</i> classifiers utilise Bayes' theorem as a formula to calculate the probability of a class based on a set of features. It is "naive" in the sense that the features are assumed to be independent of each other. There are multiple types of Naive Bayes classifiers with slightly different use-cases. The model presented here is a Gaussian Naive Bayes that assumes a normal distribution for all values of the different features.<br>
<h3>Pros:</h3>
<ol>
    <li>Efficient/fast to run</li>
    <li>Can handle larger amounts of features</li>
    <li>Good performance even with smaller training sets</li>
    <li>Works with categorical data</li>
</ol>
<h3>Cons:</h3>
<ol>
    <li>Assumes feature-independence, which is rarely the case with real world data</li>
    <li>Outcomes not present in the training data might not be recognised by the model</li>
</ol>

<h3>Implementation:</h3>

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model_GNB = GaussianNB() #Calling the model
model_GNB.fit(X_train, y_train) #Training the model
pred_GNB = model_GNB.predict(X_test) #Letting the model predict on the test set
```

```python
print('Testing accuracy: ', accuracy_score(y_test, pred_GNB))
print('Training accuracy: ', accuracy_score(y_train, model_GNB.predict(X_train)))
```
